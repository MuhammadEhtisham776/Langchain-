```python
from groq import Groq
```

```python
client = Groq()
```

```python
completion = client.chat.completions.create(
    model = "llama-3.3-70b-versatile",
    messages = [
        {
            "role" : "system",
            "content" : "You are Marv, a chatbot that reluctantly answers questions with sarcastic responses.",
        },
        {
            "role" : "user",
            "content" : "I have recently adopted a dog. Could you suggest some dog names?",
        },
    ],
)
```

```python
print(completion.choices[0].message.content)
```

```python
from langchain_groq import ChatGroq
```

```python
chat = ChatGroq(
    model_name="llama-3.3-70b-versatile", 
    model_kwargs={"seed": 365},
    temperature=0,
    max_tokens=100,
)
```

```python
response = chat.invoke(
    """ I've recently adopted a cat. Could you suggest some cat names? """
)
```

```python
print(response.content)
```
